## Down the Hill

Now that we cdan measure how wrong a set of weights is, the question now beecomes how to improve them.
We could use trial and error but as the number of weights scale up, this proves futile. So we need a way to 
work out every weight at once which reduces the loss and by how much

Picture the loss as a landscape and you are standing on this landscape in a thick fog. You can't see the bottom but you 
can feel the slope. The sensible strategy here would be to a take a step and feel the gradient of the slope under your feet and repeat.
This is gradient descent

For each parameter, gradient descent tells you how much the loss would change if you nudged that parameter up slightly. A positive value means
increasing the weight makes the loss worse, so you should decrease it. A negative value indicates the opposite.

- w <- w - $\eta $ (gradient with respect to w)
- b <- b - $\eta $ (gradient with respect to b)

$\eta $ stands for the learning rate, which sets the step size. This part is chosen and not learned.

**Deriving the gradient**

To derive the gradient, we start ou with the loss for one example from Stage 2:
$$ L = log(1 + {e^z}) - y.z$$

Differentiate with respect to z. The derivative of $log(1 + {e^z})$ is ${e^z} / {(1 + {e^z})}$ which is the same as $\sigma (z)$ (p)
and the derivative of y.z is z.
This means: $${\partial L / \partial z} = p - y $$ 

The gradient of the loss with respect to the score (z) is simply the prediction minus the true distributio (y).

Now we get back to the weights. Since z = w1x1 + w2w2 + .....b, nudging weight w changes z by x and nudging b changes z by 1.

For example: 

- ${\partial L/  \partial w} = (p - y).x $
- ${\partial L  \partial b} = (p - y) $
